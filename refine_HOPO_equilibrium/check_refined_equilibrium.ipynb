```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from matplotlib import cm
import make_pes as mpes
from scipy.optimize import minimize
```

```python
p, orders = mpes.get_coeffs_orders('equillibrium_refined_fit.res')
nonlin_p = mpes.get_non_linear_p('equillibrium_refined_fit.res')
pes = mpes.get_pes_v7_sin_sin(p, nonlin_p.copy(), orders)

print(nonlin_p)
```

```python
def pes_fitted_cis(input_geometry):
    input_geometry = np.append(input_geometry, 0)  
    geometry = np.array([input_geometry])
    return pes(geometry)

def pes_fitted_trans(input_geometry):
    input_geometry = np.append(input_geometry, 180)  
    geometry = np.array([input_geometry])
    return pes(geometry)

guess = np.array([1.68074424, 1.43555405, 0.96380778, 110.46377558, 115.67092131])
cis_minima = minimize(pes_fitted_cis, 
                  guess, 
                  method='Nelder-Mead',
                  options={"disp": False})

print('Cis Minimum:')
print(cis_minima.x)
print()

trans_minima = minimize(pes_fitted_trans, 
                  guess, 
                  method='Nelder-Mead',
                  options={"disp": False})

print('Trans Minimum:')
print(trans_minima.x)
```
