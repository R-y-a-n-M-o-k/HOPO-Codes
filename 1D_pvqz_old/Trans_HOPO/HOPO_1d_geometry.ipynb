```python
import numpy as np 
```

```python
rpo1 = 1.468331
rpo2 = 1.604292
roh = 0.967982
aopo = 110.728527
apoh = 113.054787
tau = 180 #changed for trans-HOPO

eq_vals = np.array([rpo1, rpo2, roh, aopo, apoh, tau])

#1d grids 
aopo_values = np.arange(75,179.5,1)  #angles
apoh_values = np.arange(55,180.5,1)
tau_values = np.arange(0,180.5,1)
rpo1_values = np.arange(1.22,2.005,0.01) #relative distances
rpo2_values = np.arange(1.30,2.405,0.01)
roh_values = np.arange(0.70,1.505,0.01)
```

```python
#removed tau values, has been done already
cut_values = [rpo1_values, rpo2_values, roh_values, aopo_values, apoh_values]
eq_values = [rpo1, rpo2, roh, aopo, apoh]
table1d = []
for i in range(len(cut_values)):
    
    eq_coords = np.concatenate((eq_vals[0:i], np.zeros(1), eq_vals[i+1:]))
    stack_coords = np.tile(eq_coords, (len(cut_values[i]), 1))
    stack_coords[:,i] = cut_values[i]
    table1d.append(stack_coords)

table1d = np.vstack(table1d)
print(np.shape(table1d))

points = np.array(range(len(table1d))) + 1
points = points.reshape(-1,1)
table1d = np.hstack((points, table1d))

print(table1d)
```

```python
np.savetxt('trans1d_geometries_1d.txt', table1d, 
           fmt=('%d','%8f','%8f','%8f','%8f','%8f', '%8f'), delimiter=' ')
```
