```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
pVTZ_vals = np.loadtxt('1D_pVTZ.txt', delimiter='\t')
pVQZ_vals = np.loadtxt('1D_pVQZ.txt', delimiter='\t')
trans_vals = np.loadtxt('trans_pVQZ.txt', delimiter='\t')

#print(pVTZ_vals)
#print(pVQZ_vals)
print(np.shape(pVTZ_vals))
print(np.shape(pVQZ_vals))

hartree_to_cm = 219474.6313705
minimum = np.min(pVQZ_vals[:,6])
pVTZ_vals[:,6] = (pVTZ_vals[:,6] - np.min(pVTZ_vals[:,6]))*hartree_to_cm
pVQZ_vals[:,6] = (pVQZ_vals[:,6] - np.min(pVQZ_vals[:,6]))*hartree_to_cm
trans_vals[:,6] = (trans_vals[:,6] - minimum)*hartree_to_cm
```

```python
idx = np.array([0,79,190,271,376,502,682])

for i in range(6):

    plt.figure(figsize=(10,6))
    plt.plot(pVTZ_vals[idx[i]:idx[i+1],i], pVTZ_vals[idx[i]:idx[i+1],6],color='blue')
    plt.plot(pVQZ_vals[idx[i]:idx[i+1],i], pVQZ_vals[idx[i]:idx[i+1],6], color='red')
    plt.ylabel('Relative energy')

    plt.figure(figsize=(10,6))
    plt.plot(pVQZ_vals[idx[i]:idx[i+1],i], pVTZ_vals[idx[i]:idx[i+1],6]-pVQZ_vals[idx[i]:idx[i+1],6], '.')
    plt.ylabel('pVQZ - pVTZ')

for i in range(5):
    plt.title('Trans-HOPO')
    plt.figure(figsize=(10,6))
    plt.plot(trans_vals[idx[i]:idx[i+1],i], trans_vals[idx[i]:idx[i+1],6], color='red')
    plt.ylabel('Relative energy')

```
