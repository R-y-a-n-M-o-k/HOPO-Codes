## Linear bend grid

The grid below modifies geometries, so $ \beta = 180^o $ and $\tau = 0^o$. This will be used to constraint the potential energy surface so it is unchanged with respect to torsion at the linear HOP configuration.

```python
linear_geometries = monte_carlo_HOPO(500, savetxt = False)
print(linear_geometries[:5])
```

```python
#make HOP bend linear (180), then set tau to 0 (arbitrary)
linear_geometries[:, 5] = 180.00000000
linear_geometries[:, 6] = 0.00000000

print(linear_geometries[:5])
```

```python
np.savetxt('constrain_torsion_grid.txt', 
           linear_geometries, 
           fmt=('%d','%.6f','%.6f','%.6f','%.6f','%.6f','%.6f'), 
           delimiter=' ')
```
