```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
pd.set_option('display.max_columns', 5)
energies = pd.read_csv('test_energy.txt', 
                       delim_whitespace=True,
                       names=['r1','r2','r3','aopo','apoh','tau','energy','point'])
print(energies)
```

```python
#minimum energy in hartree, convert hartree to cm^-1
min_E = -491.9191405575
hartree_to_cm = 219474.6313705

energies['energy'] = (energies['energy'] - min_E)*hartree_to_cm
print(energies)
```

```python
energies.sort_values(by='energy', axis='rows', inplace=True)
energies.to_csv("cleaned_test_energy.txt", 
                sep=" ", 
                index=False, 
                header=False, 
                float_format="%.8f")

print(energies)
```

```python
plt.figure()
plt.hist(energies['energy'], bins=100, alpha=0.6, color='b', label='Samples');
```
