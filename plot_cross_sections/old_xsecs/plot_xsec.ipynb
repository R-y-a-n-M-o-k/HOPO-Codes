```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
_ = np.loadtxt('try_exocross.xsec')
print(_)
wavenumber, cross_sec = _[:, 0], _[:, 1]
```

```python
plt.figure(figsize=(10,7))
plt.plot(wavenumber, cross_sec)
plt.yscale("log")
plt.xlabel(r"Wavenumber (cm$^{-1}$)")
plt.ylabel("Cross Section")
plt.xlim(0,5000)
plt.ylim(1e-26, 1e-17)
```
